```python
# Read JSON file from URL and print data
import requests

url = "https://www.gov.uk/bank-holidays.json"
response = requests.get(url)
data = response.json()
print (data)
```

```python
# Extract and print the dates of holidays for Northern Ireland
northern_ireland_holidays = data['northern-ireland']['events']
for holiday in northern_ireland_holidays:
    print(holiday['date'])
```

```python
# Extract holidays for England and Wales, and Scotland
england_wales_holidays = data['england-and-wales']['events']
scotland_holidays = data['scotland']['events']

# Combine holidays from England and Wales, and Scotland
other_uk_holidays = {holiday['date'] for holiday in england_wales_holidays}.union(
    {holiday['date'] for holiday in scotland_holidays}
)

# Find holidays unique to Northern Ireland
unique_northern_ireland_holidays = [
    holiday for holiday in northern_ireland_holidays if holiday['date'] not in other_uk_holidays
]

# Print unique holidays
for holiday in unique_northern_ireland_holidays:
    print(f"{holiday['title']} on {holiday['date']}")
```
