##### (1.2) Modify Programme to Deal with Header Line Separetely 

```python
# Reading CSV Data 
import csv
import os

FILENAME = 'data.csv'
DATADIR = r'C:\AlecProjects\PFDA_2025\Lab01_DataRepresentation'

file_path = os.path.join(DATADIR, FILENAME)

with open(file_path, 'rt') as fp:
    reader = csv.reader(fp, delimiter=",")
    linecount = 0
    for line in reader:
        if not linecount:
            print(f"{line}\n-------------")
        else:
            print(line)
        linecount += 1
```

##### (1.3) Modify the Programme to deal calculate the average age

```python
import csv
import os

# Calculate the average age

FILENAME = 'data.csv'
DATADIR = r'C:\AlecProjects\PFDA_2025\Lab01_DataRepresentation'

file_path = os.path.join(DATADIR, FILENAME)

with open(file_path, 'rt') as fp:
    reader = csv.reader(fp, delimiter=",")
    linecount = 0
    total_age = 0
    for line in reader:
        if not linecount:
            print(f"{line}\n-------------")
        else:
            total_age += int(line[1])  # Assuming the 'age' column is the second column
        linecount += 1

    average_age = total_age / (linecount - 1)  # Exclude the header row
    print(f"Average Age: {average_age}")
```
